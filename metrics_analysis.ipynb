```python
import os
import json
import matplotlib.pyplot as plt
import numpy as np
```

# Mask Conditioned Synthetic Images 

### KID

```python
filename = '/mnt/local/work/filippo.schiazza2/Paper/Metrics/CTR_cond/kid_results.json'
with open(filename, 'r') as f:
    kid = json.load(f)
```

```python
x = kid['N_kid']
y = kid['KID_mean']
yerr = kid['KID_std']

plt.errorbar(x, y, yerr=yerr, fmt='o')
plt.xlabel('Number of images')
plt.ylabel('KID')
plt.title('KID vs Number of images')
plt.grid()
plt.savefig('/mnt/local/work/filippo.schiazza2/Paper/Metrics/CTR_cond/KID_vs_N.png')
plt.savefig('/mnt/local/work/filippo.schiazza2/Paper/Metrics/CTR_cond/KID_vs_N.svg')
plt.show()

print(f'KID: {y[-1]:.3f} +/- {yerr[-1]:.3f}')
```

### FID infinity

```python
filename = '/mnt/local/work/filippo.schiazza2/Paper/Metrics/CTR_cond/fid_results.json'
with open(filename, 'r') as f:
    fid = json.load(f)
```

```python
# evaluate the final FID infinity and its error
num_experiments = fid['num_experiments']
values = []
for i in range(num_experiments):
    values.append(fid[str(i)]['intercept'])
    x = list(fid[str(i)].keys())[:-2]
    x = np.array(x[::-1]).astype(float)
    x = 1 / x
    y = list(fid[str(i)].values())[:-2]
    y = np.array(y[::-1])
    m = fid[str(i)]['slope']
    q = fid[str(i)]['intercept']
    plt.scatter(x, y, label='experiment {}'.format(i+1), s=10)
    x = np.insert(x, 0, 0)
    plt.plot(x, m * x + q)

plt.xlabel('1/N')
plt.ylabel('FID')
plt.title('FID vs 1/N')
plt.legend(prop={'size': 5})
# reduce the number of digits in the x-axis
plt.gca().xaxis.set_major_formatter(plt.FuncFormatter(lambda x, _: '{:.4f}'.format(x)))
plt.grid()
plt.savefig('/mnt/local/work/filippo.schiazza2/Paper/Metrics/CTR_cond/FID_vs_1_N.png')
plt.savefig('/mnt/local/work/filippo.schiazza2/Paper/Metrics/CTR_cond/FID_vs_1_N.svg')
plt.show()

mean = np.mean(values)
std = np.std(values)

print('FID infinity: {:.2f} +/- {:.2f}'.format(mean, std))
```
